```python
import numpy as np
import matplotlib.pyplot as mplt
```

```python
'''
Vamos a construir una simulación sencilla para entender cómo se forma una distribución muestral de la media. Esto 
permitirá ver cómo las medias de muchas muestras se distribuyen, incluso si la población no es normal.
'''

# Crear una población con distribución uniforme
poblacion = np.random.uniform(0, 100, 10000)

# Parámetros de muestreo
n_muestras = 1000     # Número de muestras
tamano_muestra = 30   # Tamaño de cada muestra

# Almacenar las medias muestrales
medias_muestrales = []

for _ in range(n_muestras):
    muestra = np.random.choice(poblacion, tamano_muestra, replace=False)
    media = np.mean(muestra)
    medias_muestrales.append(media)

# Graficar la distribución muestral de la media
mplt.hist(medias_muestrales, bins=30, edgecolor='black', color= 'green', alpha=0.7)
mplt.title('Distribución Muestral de la Media (n=30)')
mplt.xlabel('Media Muestral')
mplt.ylabel('Frecuencia')
mplt.grid(True)
mplt.show()
```
