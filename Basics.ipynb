```python
def my_function(x):
  return x[::-1]

mytxt = my_function("I wonder how this text looks like backwards")

print(mytxt)
```

```python

texto = "Mi número es 809-555-1234 y el de mi amigo es 829-333-5678"
patron = r"\d{3}-\d{3}-\d{4}"

resultados = re.findall(patron, texto)
print(resultados)
```

```python
#Two references to the same object 
a = [1, 2, 3]
b = a

print(a is b)  # True, both variables point to the same list object
print(a == b)  # True, both lists have the same content 

print(b)  # Output: [1, 2, 3]

a.append(4)
print(b)  # Output: [1, 2, 3, 4]    
```
